#### Cell 2: `from matplotlib import pyplot as plt`
Baris ini mengimpor modul `pyplot` dari pustaka `matplotlib` dan memberinya alias `plt`. `matplotlib.pyplot` adalah koleksi fungsi yang membuat `matplotlib` bekerja seperti MATLAB. Ini digunakan untuk **memvisualisasikan gambar dan grafik**.

```python
from PIL import Image
from PIL import ImageEnhance
from matplotlib import pyplot as plt

# Opens the image file
image = '/content/drive/MyDrive/Colab Notebooks/MyImage/AgusPop_kecil.jpg'
image = Image.open(image)
# shows image in image viewer
#img = plt. imread('image.png')
plt.imshow(image)
plt.show()
#image.show()


# Enhance Brightness
enhancer = ImageEnhance.Brightness(image)
img_brighteted = enhancer.enhance(1.2)
img_brighteted.save('/content/drive/MyDrive/Colab Notebooks/MyImage/AgusPop_kecil_Brigt.jpg')
plt.imshow(img_brighteted)
plt.show()
```

#### Cell 3: Memuat, Menampilkan, dan Mencerahkan Gambar Awal

Blok kode ini melakukan beberapa langkah penting:
1.  **Mengimpor Pustaka**: Mengimpor `Image` dan `ImageEnhance` dari `PIL` (Pillow) untuk manipulasi gambar, serta `pyplot` dari `matplotlib` untuk menampilkan gambar.
2.  **Memuat Gambar**: Membuka file gambar `'AgusPop_kecil.jpg'` dari Google Drive Anda ke dalam variabel `image`.
3.  **Menampilkan Gambar Asli**: Menggunakan `plt.imshow(image)` dan `plt.show()` untuk menampilkan gambar asli sebelum dimanipulasi.
4.  **Mencerahkan Gambar**: Membuat objek `ImageEnhance.Brightness` dari gambar asli, lalu mencerahkannya sebesar `1.2` (120% dari kecerahan asli). Hasilnya disimpan di `img_brighteted`.
5.  **Menyimpan Gambar yang Dicerekahkan**: Menyimpan gambar yang sudah dicerahkan ke lokasi baru di Google Drive Anda.
6.  **Menampilkan Gambar yang Dicerekahkan**: Menampilkan gambar yang telah dicerahkan.

```python
curr_bri = ImageEnhance.Brightness(image)
new_bri = 1.5

# Brightness enhanced by a factor of 1.5
img_brightened = curr_bri.enhance(new_bri)

# shows updated image in image viewer
plt.imshow(img_brightened)
```

#### Cell 5: Meningkatkan Kecerahan Gambar

Blok kode ini secara spesifik berfokus pada peningkatan kecerahan:
1.  **Menginisialisasi Enhancer Kecerahan**: Membuat objek `ImageEnhance.Brightness` dari gambar asli (variabel `image`).
2.  **Menentukan Faktor Kecerahan**: Mengatur `new_bri` menjadi `1.9`, yang berarti gambar akan 190% lebih cerah dari aslinya.
3.  **Menerapkan Kecerahan**: Menggunakan metode `enhance(new_bri)` untuk mencerahkan gambar.
4.  **Menampilkan Hasil**: Menampilkan gambar `img_brightened` yang telah ditingkatkan kecerahannya.

```python
curr_con = ImageEnhance.Contrast(image)
new_con = 0.3

# Contrast enhanced by a factor of 0.3
img_contrasted = curr_con.enhance(new_con)

# shows updated image in image viewer
plt.imshow(img_contrasted)
```

#### Cell 6: Meningkatkan Kontras Gambar

Blok ini mirip dengan peningkatan kecerahan, tetapi untuk kontras:
1.  **Menginisialisasi Enhancer Kontras**: Membuat objek `ImageEnhance.Contrast` dari gambar asli.
2.  **Menentukan Faktor Kontras**: Mengatur `new_con` menjadi `4`, yang berarti gambar akan memiliki kontras 4 kali lipat dari aslinya.
3.  **Menerapkan Kontras**: Menggunakan `enhance(new_con)` untuk meningkatkan kontras.
4.  **Menampilkan Hasil**: Menampilkan gambar `img_contrasted` yang kontrasnya telah ditingkatkan.

```python
# Enhance Sharpness
curr_sharp = ImageEnhance.Sharpness(image)
new_sharp = 8.3

# Sharpness enhanced by a factor of 8.3
img_sharped = curr_sharp.enhance(new_sharp)

# shows updated image in image viewer
plt.imshow(img_sharped)
```

#### Cell 7: Meningkatkan Ketajaman Gambar (Sharpness)

Blok ini fokus pada ketajaman gambar:
1.  **Menginisialisasi Enhancer Ketajaman**: Membuat objek `ImageEnhance.Sharpness` dari gambar asli.
2.  **Menentukan Faktor Ketajaman**: Mengatur `new_sharp` menjadi `10`, yang akan membuat gambar 10 kali lebih tajam.
3.  **Menerapkan Ketajaman**: Menggunakan `enhance(new_sharp)` untuk mempertajam gambar.
4.  **Menampilkan Hasil**: Menampilkan gambar `img_sharped` yang ketajamannya telah ditingkatkan.

```python
curr_col = ImageEnhance.Color(image)
new_col = 2.5

# Color level enhanced by a factor of 2.5
img_colored = curr_col.enhance(new_col)

# shows updated image in image viewer
plt.imshow(img_colored )
```

#### Cell 8: Meningkatkan Saturasi Warna Gambar (Color)

Blok kode terakhir ini berkaitan dengan saturasi warna:
1.  **Menginisialisasi Enhancer Warna**: Membuat objek `ImageEnhance.Color` dari gambar asli.
2.  **Menentukan Faktor Warna**: Mengatur `new_col` menjadi `11`, yang berarti gambar akan memiliki saturasi warna 11 kali lipat dari aslinya.
3.  **Menerapkan Warna**: Menggunakan `enhance(new_col)` untuk meningkatkan saturasi warna.
4.  **Menampilkan Hasil**: Menampilkan gambar `img_colored` yang saturasi warnanya telah ditingkatkan.
