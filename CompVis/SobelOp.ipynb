```python
from typing import TypeVar
import numpy as np
import numpy.typing as npt
import cv2
from google.colab.patches import cv2_imshow
import matplotlib.pyplot as plt
from numpy import asarray
```

```python
T = TypeVar("T", bound=npt.NBitBase)
```

```python
def fsobel(image:'np.ndarray[T][T]'):
    Gx = [[1, 0, -1],
		    [2, 0, -2],
		    [1, 0, -1]]

    Gy = [[1, 2, 1],
		    [0, 0, 0],
		    [-1, -2, -1]]
    image = np.asarray(image)
    axes = image.shape
    rows = axes[0]
    cols = axes[1]
    mag = np.zeros_like(image) # to store output differential/gradient
    i=j=0
    for i in range (rows)[1:-2]:
      for j in range (cols)[1:-2]:
        S1 = np.sum(np.multiply(Gx, image[i:i+3, j:j+3]))
        S2 = np.sum(np.sum(np.multiply(Gy, image[i:i+3, j:j+3])))
        mag[i+1, j+1] = np.sqrt(np.sum([np.square(S1), np.square(S2)]))
        j = j+1
      i = i+1
    mag *= 255.0 / mag.max()
    output_image = mag
    return(output_image)
```
