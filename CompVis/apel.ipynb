````python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow


# Membaca gambar
img = cv2.imread('/content/drive/MyDrive/Colab Notebooks/MyImage/apel.jpg')
cv2_imshow(img)

````

````python
image_resized_down = cv2.resize(img,(100,100))
cv2_imshow(image_resized_down)
````

````python
import numpy as np
import matplotlib.pyplot as plt

hist_blue_channel = cv2.calcHist([image_resized_down],[0],None,[256],[0,256])
hist_green_channel = cv2.calcHist([image_resized_down],[1],None,[256],[0,256])
hist_red_channel = cv2.calcHist([image_resized_down],[2],None,[256],[0,256])


plt.plot(hist_blue_channel, color='blue', label='Blue Channel')
plt.plot(hist_green_channel, color='green', label='Green Channel')
plt.plot(hist_red_channel, color='red', label='Red Channel')
plt.legend()
plt.show()
````

````python
import cv2
# Konversi dari BGR ke HSV
img_hsv = cv2.cvtColor(image_resized_down, cv2.COLOR_BGR2HSV)
cv2_imshow(img_hsv)

# 1. Hitung Histogram Warna 3D
# Menentukan jumlah bins (kotak) untuk masing-masing saluran H, S, dan V
# Contoh: 8 bins untuk Hue, 2 bins untuk Saturation, 2 bins untuk Value
hist_3d = cv2.calcHist([img_hsv], [0, 1, 2], None, [8, 2, 2], [0, 180, 0, 256, 0, 256])

# 2. RATAKAN (Flatten) matriks 3D menjadi Vektor 1D
# Ini akan menghasilkan 8 x 2 x 2 = 32 elemen angka dalam satu baris
vektor_hist_1d = hist_3d.flatten()

# Cetak ukuran vektor dan isinya di terminal
print("Ukuran Vektor 1D:", vektor_hist_1d.shape)
print("Isi Vektor 1D:\n", vektor_hist_1d)

# Tampilkan grafik histogram 1D menggunakan Matplotlib
plt.figure(figsize=(10, 4))
plt.bar(range(len(vektor_hist_1d)), vektor_hist_1d, color='blue', edgecolor='black')
plt.title("Visualisasi Vektor Color Histogram 1D (32 Bins)")
plt.xlabel("Indeks Vektor (Bin)")
plt.ylabel("Frekuensi Jumlah Piksel")
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.show()
````

````python
# Konversi dari BGR ke RGB
img_rgb = cv2.cvtColor(image_resized_down, cv2.COLOR_BGR2RGB)
cv2_imshow(img_rgb)


# Menampilkan numerisasi matriks
# Kita gunakan np.array() untuk memastikan formatnya adalah matriks numerik
matriks_pixel = np.array(img_rgb)
````
