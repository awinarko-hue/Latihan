```python
# Data transaksi
transaksi = [
    ["Teh", "Gula", "Susu"],
    ["Teh", "Gula"],
    ["Teh", "Susu"],
    ["Susu", "Roti"],
    ["Roti", "Gula", "Susu"],
    ["Teh", "Roti", "Gula"],
    ["Gula", "Susu"],
    ["Teh", "Susu", "Gula"]
]

# Hitung jumlah transaksi
total_transaksi = len(transaksi)

# Hitung support item tunggal
items = ["Teh", "Gula", "Susu", "Roti"]
support_item = {}

for item in items:
    count = sum([1 for t in transaksi if item in t])
    support_item[item] = count / total_transaksi

print("Support Item Tunggal:")
for k,v in support_item.items():
    print(f"{k}: {v:.3f}")

# Hitung support dan confidence untuk kombinasi 2-item
from itertools import combinations

kombinasi = list(combinations(items, 2))
hasil = []

for (a,b) in kombinasi:
    count_ab = sum([1 for t in transaksi if a in t and b in t])
    support_ab = count_ab / total_transaksi
    confidence_a_b = support_ab / support_item[a] if support_item[a] > 0 else 0
    confidence_b_a = support_ab / support_item[b] if support_item[b] > 0 else 0
    hasil.append({
        "Kombinasi": f"{a} & {b}",
        "Support": support_ab,
        "Confidence A→B": confidence_a_b,
        "Confidence B→A": confidence_b_a,
        "Support×Confidence A→B": support_ab * confidence_a_b,
        "Support×Confidence B→A": support_ab * confidence_b_a
    })

# Tampilkan hasil
import pandas as pd
df = pd.DataFrame(hasil)
print("\nHasil Kombinasi 2-Item:")
print(df)

# Cari kombinasi terbaik berdasarkan Support×Confidence
best = df.loc[df[["Support×Confidence A→B","Support×Confidence B→A"]].max(axis=1).idxmax()]
print("\nKombinasi terbaik berdasarkan Support×Confidence:")
print(best)
```
