# 1. Install dan inisiasi pustaka Pytholog

```python
!pip install pytholog
import pytholog as pl

kb = pl.KnowledgeBase("evaluasi_probabilistik")
```

# 2. Input fakta data uji

```python
fakta_data = [
    ("cisro", 20, 2550, 20),
    ("huamei", 19.8, 2300, 18),
    ("asusa", 14.5, 2480.5, 23),
    ("terpal", 23, 2450.5, 13),
    ("ubigoreng", 18.5, 2430, 20),
    ("tplim", 12.8, 2420, 19),
    ("zti", 24.5, 2420.5, 18),
]
```

# 3. Kelola Fakta dengan perhitungan


*   Perhitungan probabilitas adalah proses komputasi yang dilakukan menggunakan Python untuk mendapatkan nilai (diluar pustaka pytholog). Ini adalah input ke Pytholog
*   Aturan fakta adalah cara  menyimpan dan merepresentasikan hasil perhitungan tersebut (beserta data asli) dalam format yang dapat dipahami dan dikueri oleh Pytholog. Ini adalah cara Pytholog mengetahui tentang data dan hasil probabilitas tersebut.



```python
facts = []
for nama, power, frek, bw in fakta_data:
    facts.append(f"fakta({nama}, {power}, {frek}, {bw})")

    prob_power = 1.0 if power <= 20 else max(0.0, 1 - (power - 20) / 10)
    prob_frek = 1.0 if 2400 <= frek <= 2500 else 0.0
    prob_bw = 1.0 if bw <= 20 else max(0.0, 1 - (bw - 20) / 10)
    prob_final = round((prob_power + prob_frek + prob_bw) / 3, 3)
    status = "lolos" if prob_final >= 0.8 else "gagal"

    facts.append(f"prob_power({nama}, {prob_power})")
    facts.append(f"prob_frek({nama}, {prob_frek})")
    facts.append(f"prob_bw({nama}, {prob_bw})")
    facts.append(f"prob_final({nama}, {prob_final})")
    facts.append(f"status({nama}, {status})")

kb(facts)
```

# 4. Helper query

```python

def show_query(expr, title):
    results = kb.query(pl.Expr(expr))
    print(f"=== {title} ===")
    if not results:
        print("  <no result>")
        return

    if isinstance(results, list) and all(isinstance(r, dict) for r in results):
        for row in results:
            if "Nama" in row and "Prob" in row:
                print(f"  {row['Nama']}: {row['Prob']}")
            elif "Nama" in row and "Status" in row:
                print(f"  {row['Nama']}: {row['Status']}")
            else:
                print("  " + ", ".join(f"{k}: {v}" for k, v in row.items()))
    else:
        if isinstance(results, list):
            for item in results:
                print(f"  {item}")
        else:
            print(f"  {results}")
    print()
```

# 5. Query hasil

```python
show_query("prob_power(Nama, Prob)", "Probabilitas Power")
show_query("prob_frek(Nama, Prob)", "Probabilitas Frekuensi")
show_query("prob_bw(Nama, Prob)", "Probabilitas BW")
show_query("prob_final(Nama, Prob)", "Probabilitas Final")
show_query("status(Nama, Status)", "Status Akhir")
```
